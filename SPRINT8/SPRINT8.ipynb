### » Corregir possibles problemes de codificació i accentuació en els noms de les columnes i en els valors de text, si escau.

Función **corregir_codificacion** comprueba si el valor es una cadena. Si lo es, intenta recodificarla desde latin1 a utf‑8 para corregir caracteres extraños. Si la conversión falla, devuelve el texto original. Si el valor no es una cadena, lo retorna sin modificar.

```python
def corregir_codificacion(valor):
  if isinstance(valor, str):
    try:
      return valor.encode('latin1').decode('utf-8')
    except (UnicodeEncodeError, UnicodeDecodeError):
      return valor
  return valor
```

 Aplicamos la función **corregir_codificacion** a cada nombre de columna del DataFrame. Recorre todas las columnas y devuelve una lista con los nombres ya corregidos, sustituyendo caracteres mal codificados.

```python
df.columns = [corregir_codificacion(columna)
  for columna in df.columns]
```

Identificamos todas las columnas de tipo texto del DataFrame y aplicamos la función **corregir_codificacion** a cada una. Así corrige caracteres mal codificados dentro de las celdas, limpiando valores corruptos y dejando todas las columnas de texto con codificación coherente.

```python
columnas_text = df.select_dtypes(include="object").columns

for columna in columnas_text:
  df[columna] = df[columna].apply(corregir_codificacion)
```

```python
def explorar_dataframe(df, carpeta="plots"):
    if not os.path.exists(carpeta):
        os.makedirs(carpeta)

    for col in df.columns:
        serie = df[col]

        plt.figure(figsize=(8, 5))

        if serie.dtype == "float64" or serie.dtype == "int64":

            plt.subplot(1, 2, 1)
            sns.histplot(serie.dropna(), kde=True)
            plt.title(f"Histograma de {col}")

            plt.subplot(1, 2, 2)
            sns.boxplot(x=serie.dropna())
            plt.title(f"Boxplot de {col}")

        elif serie.dtype == "object":

            conteo = serie.value_counts().head(10)
            sns.barplot(x=conteo.values, y=conteo.index)
            plt.title(f"Valores más frecuentes de {col}")

        elif str(serie.dtype).startswith("datetime"):

            años = serie.dt.year.value_counts().sort_index()
            sns.barplot(x=años.index, y=años.values)
            plt.title(f"Distribución por año de {col}")
            plt.xticks(rotation=45)

        elif serie.dtype == "bool":

            conteo = serie.value_counts()
            sns.barplot(x=conteo.index.astype(str), y=conteo.values)
            plt.title(f"Distribución de valores booleanos en {col}")

        else:

            plt.text(0.5, 0.5, f"No se puede graficar la columna {col}", ha="center")

        ruta = os.path.join(carpeta, f"{col}.png")
        plt.tight_layout()
        plt.savefig(ruta)
        plt.close()

        print(f"Gráficos generados y guardados en la carpeta: {carpeta}")
```


    Función que calcula una ruta aproximada entre dos ciudades
    usando una estrategia greedy: siempre ir a la ciudad no visitada
    más cercana hasta llegar a la ciudad destino.
    

  BLOQUE 1: Inicialización ---
  Creamos una lista para guardar el orden de ciudades visitadas.

  Conjunto para marcar ciudades ya visitadas.

Ciudad actual desde la que buscamos la siguiente.

 Distancia total acumulada.

  BLOQUE 2: Bucle principal ---
  Continuamos hasta llegar a la ciudad destino.

Distancias desde la ciudad actual a todas las demás.

Filtramos ciudades no visitadas.

Elegimos la ciudad más cercana.

Actualizamos ruta y distancia.

Marcamos como visitada.

Avanzamos a la siguiente ciudad.

  BLOQUE 3: Resultado ---
Devolvemos la ruta completa y la distancia total recorrida.

```python
def ruta_greedy(origen, destinacio, df_dist):

    ruta = [origen]

    visitadas = set([origen])

    actual = origen

    distancia_total = 0

    while actual != destinacio:

        distancias = df_dist.loc[actual]

        no_visitadas = distancias.drop(labels=list(visitadas))

        siguiente_ciudad = no_visitadas.idxmin()
        distancia_siguiente = no_visitadas.min()

        ruta.append(siguiente_ciudad)
        distancia_total += distancia_siguiente

        visitadas.add(siguiente_ciudad)

        actual = siguiente_ciudad

    return ruta, distancia_total
```
