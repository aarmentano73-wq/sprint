### **1-FUNCIÓN obtenir_dades**
>* Solicitamos datos al usuario.  

>* Retorno de los datos que necesitamos usarlos mas tarde

```python
def obtenir_dades():
  nom_cognoms = input("Introdueix el teu nom i cognoms: ")
  telefon = int(input("Introdueix el teu numero de telefon: "))
  any_experiencia = int(input("Introdueix els anys d'experiencia: "))
  nivell_catala = int(input("Introdueix el teu nivell de català: "))
  nivell_excel =  int(input("Introdueix el teu nivell de excel: "))
  titol_academic = int(input("Introduieix el teu nivell académic: "))

  #retorno como diccionario: clave : valor

  return {"nom_cognoms": nom_cognoms,
          "telefon": telefon,
          "any_experiencia": any_experiencia,
          "nivell_catala": nivell_catala,
          "nivell_excel": nivell_excel,
          "titol_academic": titol_academic
          }
```

### **2-FUNCIÓN validar dades**

Validación de datos correctos:

>* Nombre y apellidos (strings)

>* Telefono (número entero y 9 digitos)

>* Años de experiencia (número entero, max 2 digitos)

>* Nivel de catalan (numero entero [0,1,2,3], 1 digito)

>* Nivel de Excel (numero entero [0,1,2,3] , 1 digito)

>* Retorno de valores.

>* Si la validación ha sido incorrecta:  

>* Devolver tipo de error

>* Imprimir mensaje


```python
def validar_dades(candidat):


    if len(candidat["nom_cognoms"]) == 0:
        print("El nom i cognoms no poden estar buits.")
        return False

    if len(str(candidat["telefon"])) != 9:
        print("El telèfon ha de tenir 9 dígits.")
        return False

    if float(candidat["any_experiencia"]) < 0:
        print("Els anys d'experiència no poden ser negatius.")
        return False

    if not (0 <= candidat["nivell_catala"] <= 3):
        print("El nivell de català ha d'estar entre 0 i 3.")
        return False

    if not (0 <= candidat["nivell_excel"] <= 3):
        print("El nivell d'Excel ha d'estar entre 0 i 3.")
        return False

    if not (0 <= candidat["titol_academic"] <= 3):
        print("La titulació ha d'estar entre 0 i 3.")
        return False

    return True
```

### **3-FUNCION cálculo puntuación**

>* Cálculo de la puntuación total del candidato aplicando la formula indicada.
>* Recibe los valores de anys_experiencia, nivell_catala, nivell_excel, nivell_titulacio
>* Devuelve puntuación total.

```python
def calcular_puntuacio(anys_experiencia, nivell_catala, nivell_excel, nivell_titulacio):
   puntuacio = (1.5*anys_experiencia) + nivell_catala + nivell_excel + nivell_titulacio

   return puntuacio
```

### **4-FUNCION clasificar candidato**

Mediante el control de flujo clasifica cada entrada (puntuacio del candidato) en 3 posibles categorias:

>* Prioritat alta

>* Candidat adecuat

>* No prioritari

```python
def classificar_candidat(puntuacio):
  if puntuacio >= 12:
    prioritat = "Alta"
  elif puntuacio >= 8:
    prioritat = "Adecuat"
  else:
    prioritat = "NO prioritari"
  return prioritat

```

### **5-FUNCION muestra candidato inscrito**  

Muestra la puntuación y clasificación  del candidato que ha ingresado los datos.

```python
def mostra_punt_ind(puntuacio, categoria):
  print("La puntuació del candidat és: ", puntuacio)
  print("La posició del candidat és: ", categoria)
```

### **6-FUNCION ordenación candidatos**
 Ordenamos los candidatos. Priorizar "key = lambda" por el valor de la clave puntuacion. Orden "reverse = True.

```python
def ordenar_candidats(candidats):

    candidats_ordenats = sorted(candidats, key=lambda candidat: candidat["puntuacio"], reverse=True )

    return candidats_ordenats
```

### **7-FUNCION muestra informe final**

Muestra el informe de los siguientes datos: los valores de las claves

>* Nombre y apellidos
>* Telefono
>* Puntuación
>* Categoria
>* Nivel Catalan
>* Nivel Excel
>* Titulacion academica
>* Años experiencia

```python
def mostra_informe(candidats):

    for candidat in candidats:
        print("Nom i cognoms:", candidat["nom_cognoms"])
        print("Telèfon:", candidat["telefon"])
        print("Anys d'experiència:", candidat["any_experiencia"])
        print("Nivell de català:", candidat["nivell_catala"])
        print("Nivell d'Excel:", candidat["nivell_excel"])
        print("Titulació acadèmica:", candidat["titol_academic"])
        print("Puntuació:", candidat["puntuacio"])
        print("Categoria:", candidat["categoria"])
        print("----------------------------")
```

### **8-FUNCION main()**


```python
def main_seleccio():

    candidats = []

    while len(candidats) < 10:
        candidat = obtenir_dades()
        if validar_dades(candidat):

            puntuacio = calcular_puntuacio(
                candidat["any_experiencia"],
                candidat["nivell_catala"],
                candidat["nivell_excel"],
                candidat["titol_academic"])

            categoria = classificar_candidat(puntuacio)

            candidat["puntuacio"] = puntuacio
            candidat["categoria"] = categoria

            candidats.append(candidat)
            mostra_punt_ind(puntuacio, categoria)

        else:
            print("Torna a introduir les dades.")

    candidats_ordenats = ordenar_candidats(candidats)
    mostra_informe(candidats_ordenats)

seleccio_candidats = main_seleccio()
```

### **1-FUNCIÓN  obtener segmento y factor**

Consulta un diccionario y devuelve el segmento de cliente y el factor de riesgo asociado.

```python


def obtenir_segment_i_factor(codi_comportament):                #La funcion recibe un parametro "codi_comportament"

                                                                #Cada codigo (0-4) relaciona 2 datos: "segment" y "factor_codi"

    comportaments = {
        0: {"missatge": "Navega però no afegeix productes", "factor_codi": 0.10},
        1: {"missatge": "Afegeix productes però no arriba al checkout","factor_codi": 0.25},
        2: {"missatge": "Arriba al checkout però no paga","factor_codi": 0.50},
        3: {"missatge": "Cistella de compra abandonada fa menys de 24 h","factor_codi": 0.75},
        4: {"missatge": "Cistella de compra abandonada fa més de 24 h","factor_codi": 0.90}
    }

    return comportaments[codi_comportament]

#La función recibe un código de comportamiento y retorna el diccionario asociado a ese código dentro del diccionario comportaments.
#Este diccionario contiene los valores de las claves "segment" y "factor_codi" correspondientes al código introducido
```

### **2-FUNCION obtener factor edad**

Consulta un diccionario con los factores de cada grupo de edad y devuelve el valor correspondiente al grupo recibido como parametro para calcular el riesgo final

```python
def obtenir_factor_edat(grup_edat):
    factors_edat = {
        "18-25": 1.20,
        "26-40": 1.00,
        "41-60": 0.85,
        "60+": 0.70
    }
    return factors_edat[grup_edat]

```

### **3-FUNCION calcular riesgo**

Calculo de riesgo

```python
def calcular_risc(factor_codi, factor_edat):
    return factor_codi * factor_edat
```

### **4-FUNCION procesar clientes**

Recorre todos los clientes, obtiene el segmento y los factores. Calcula el riesgo final y crea una nueva lista (clients_processats) con la información de cada cliente.

```python
def processar_clients(clients):
    clients_processats = []

    for client in clients:
# Obtener el segmento y el factor del comportamiento
        dades_comportament = obtenir_segment_i_factor(client["codi_comportament"])
# Obtener el factor de edad
        factor_edat = obtenir_factor_edat(client["grup_edat"])
# Calcular el riesgo final
        risc_final = calcular_risc(dades_comportament["factor_codi"],factor_edat)
# Crear un nuevo diccionario con toda la información
        nou_client = {
            "id_client": client["id_client"],
            "nom": client["nom"],
            "edat": client["edat"],
            "grup_edat": client["grup_edat"],
            "codi_comportament": client["codi_comportament"],
            "missatge": dades_comportament["missatge"],
            "factor_codi": dades_comportament["factor_codi"],
            "factor_edat": factor_edat,
            "risc_final": risc_final
        }

        clients_processats.append(nou_client)

    return clients_processats
```

### **5-FUNCION muestra diccionario**

La función recorre la lista clients_processats e imprime cada diccionario con la información de todos los clientes.

```python
def mostrar_diccionari(clients_processats):

    for client in clients_processats:
        print(client)
```

### **6-FUNCION muestra respuesta**

La función analiza la lista de clientes procesados para identificar el cliente:
>* con mayor riesgo
>* comparar riesgos por edad
>* detectar clientes con poco interés
>* mostrar el más rentable.

```python
def mostrar_respostes(clients_processats):
    print("\n")

    # 1. Cliente con mayor riesgo
    client_mes_risc = max(clients_processats, key=lambda client: client["risc_final"])

    print("1. Client que s'ha de contactar primer en una campanya de retargeting:")
    print(client_mes_risc["nom"], "-", client_mes_risc["risc_final"])

    # 2. Diferencias por edad
    print("\n2. Diferències segons l'edat:")

    for client in clients_processats:
        if client["codi_comportament"] == 4:
            print(client["nom"], "-", client["grup_edat"], "-", client["risc_final"])
    # 3. Menor interés
    print("\n3. Clients amb menys interès:")

    for client in clients_processats:
        if client["risc_final"] <= 0.25:
            print(client["nom"], "-", client["risc_final"])
    # 4. Cliente más rentable
    print("\n4. Client més rendible:")

    print(client_mes_risc["nom"], "-", client_mes_risc["missatge"])
```

###**7-FUNCION main()**

```python


def main_marketing():
    # Procesar los clientes
    clients_processats = processar_clients(clients)

    # Mostrar el resultado
    mostrar_diccionari(clients_processats)

    # Responder a las preguntas de marketing
    mostrar_respostes(clients_processats)

main_marketing()
```

### **1-FUNCION convertir pacientes**
La función convertir_pacients() transforma un registro de un paciente, almacenado como una cadena de texto,
 en datos estructurados: nombre, edad y síntomas

```python
def convertir_pacients(pacient):

  #divide la cadena cada vez que aparece "|" y la convierte en una lista
    dades = pacient.split("|")
  # se accede a los elementos de la lista por indice
    nom = dades[0]
    edat = int(dades[1])
    simptomes = dades[2]
    return nom, edat, simptomes


for pacient in pacients:
    nom, edat, simptomes = convertir_pacients(pacient)
    print(nom, edat, simptomes)


```

### **2-FUNCION natejar_simptomes**
 Normaliza y limpia los datos de los simptomas. Convertir a minusculas, remplazar espacios internos por"_" y elimina los espacios en blanco que hay

```python
def netejar_simptomes(simptomes):

    llista_simptomes = simptomes.split(",") #divide la cadena en una lista

    simptomes_nets = []

    for simptom in llista_simptomes:
        simptom = simptom.strip() #elimina los espacios al principio y al final
        simptom = simptom.lower() #transforma a minusculas
        simptom = simptom.replace(" ", "_") #remplaza los espacios en _
        simptomes_nets.append(simptom) #añade los elementos a la lista

    return simptomes_nets
```

### **3-FUNCION calcular risgo**
La funcion calcular_risc() calcula el riesgo total sumando el nivel de riesgo asociado a cada uno de sus sintomas.
Se recorre la lista de sintomas 'normalizada' consulta el valor de cada uno en el diccionario nivells_risc y
acumula el resultado en la variable risc_total

```python
def calcular_risc(simptomes, nivells_risc):
  risc_total = 0

  for simptom in simptomes:
    risc_total += nivells_risc[simptom]

  return risc_total

#Comprobación

simptomes = ["tos", "fiebre_alta", "dolor_pecho"]
risc = calcular_risc(simptomes, nivells_risc)
print(risc)
```

### **4-FUNCION procesar pacientes**
 La función processar_pacients:

>* convierte el string en datos utiles
>* limpia "simptomes"
>* calcula el "risc_total"
>* crea un diccionario para cada paciente
>* retorna una lista de pacientes procesada

```python
def processar_pacients(pacients, nivells_risc):

    pacients_processats = []

    for pacient in pacients:

        nom, edat, simptomes = convertir_pacients(pacient)
        simptomes = netejar_simptomes(simptomes)
        risc_total = calcular_risc(simptomes, nivells_risc)
        pacient_processat = {
            "nom": nom,
            "edat": edat,
            "simptomes": simptomes,
            "risc_total": risc_total
        }
        pacients_processats.append(pacient_processat)

    return pacients_processats
```

### **5-FUNCION ordenar_pacientes**
La función ordenar_pacients() ordena la lista de pacientes según el valor de risc_total, de mayor a menor.
 Para ello utiliza la función sorted(), indicando como criterio de ordenación el riesgo total de cada paciente (risc_total).
  Finalmente devuelve la lista ordenada, permitiendo establecer la prioridad de atención médica.

```python
def ordenar_pacients(pacients_processats):

    pacients_ordenats = sorted(pacients_processats, key=lambda pacient: pacient["risc_total"], reverse=True)

    return pacients_ordenats

# Comprobacion
# Lista de prueba

pacients_processats = [
    {"nom": "Jordi", "risc_total": 2},
    {"nom": "Maria", "risc_total": 8},
    {"nom": "Luis", "risc_total": 2},
    {"nom": "Anna", "risc_total": 4},
    {"nom": "Sara", "risc_total": 9}
]

# Llamar a la función

pacients_ordenats = ordenar_pacients(pacients_processats)

# Mostrar el resultado

for pacient in pacients_ordenats:
    print(pacient)
```

### **2-Función para clasificar al cliente**
Esta función utiliza el código del cliente para buscar directamente la información en el diccionario nivells_seguro.

```python
def classificar_client(client, nivells_seguro):

    dades = nivells_seguro[client["codi"]]         #busca en el diccionario nivells_seguro la información correspondiente al código del cliente.

    client["categoria"] = dades["categoria"]       #añade una clave al diccionario del cliente "categoria"
    client["preu_base"] = dades["preu_base"]       #idem "preu_base"
    client["probabilitat_frau"] = dades["fraude"]  #idem "fraude"

    return client
```

### **3-Calcular el precio final**
Cálculo segun la formula:

 preu_final = preu_base × (1 + edat/100)




```python
def calcular_preu(client):

    preu = client["preu_base"] * (1 + client["edat"] / 100)

    client["preu_final"] = round(preu,2)

    return client
```

### **4-Procesar todos los clientes**
La función processar_clients() recorre la lista de clientes, clasifica cada cliente, calcula el precio final de la póliza y guarda los resultados en una nueva lista procesada.

```python
def processar_clients(clients, nivells_seguro):

    resultat = []

    for client in clients:                                #Recorremos toda la lista
        client = classificar_client(client, nivells_seguro)
        client = calcular_preu(client)

        resultat.append(client)

    return resultat
```

###**6-Buscar el mayor fraude**
 La funcion major_frau() empieza suponiendo que Ana tiene el mayor fraude.Después compara:
Ana con Carlos
Carlos con Isabel
Isabel con Jorge
Isabel con Marta
Al final devuelve Isabel.

```python
def major_frau(clients):
    major = clients[0]

    for client in clients:
        if client["probabilitat_frau"] > major["probabilitat_frau"]:
            major = client

    return major







```

###**7-Buscar el precio más alto**
La función recorre la lista de clientes, compara el precio final de cada uno y devuelve el cliente que tiene el precio final más alto.

```python
def major_preu(clients):

    major = clients[0]

    for client in clients:
        if client["preu_final"] > major["preu_final"]:
            major = client

    return major
```

###**8-Ordenar por fraude**
 La función ordenar_frau ordena la lista de clientes de mayor a menor probabilidad de fraude. Compara cada cliente con el resto e intercambia sus posiciones cuando encuentra una probabilidad de fraude superior.

```python
def ordenar_frau(clients):

    for i in range(len(clients)-1):
        for j in range(i+1, len(clients)):
            if clients[i]["probabilitat_frau"] < clients[j]["probabilitat_frau"]:
                auxiliar = clients[i]
                clients[i] = clients[j]
                clients[j] = auxiliar

    return clients
```

###**9-Ordenar por precio**
La función ordena la lista de clientes de mayor a menor según el precio final de la póliza. Compara los precios finales e intercambia la posición de los clientes cuando encuentra un valor superior.

```python
def ordenar_preu(clients):

    for i in range(len(clients)-1):
        for j in range(i+1, len(clients)):
            if clients[i]["preu_final"] < clients[j]["preu_final"]:

                auxiliar = clients[i]
                clients[i] = clients[j]
                clients[j] = auxiliar

    return clients
```

###**10-Análisis**
 La función analisis() analiza la lista de clientes, identifica el cliente con mayor probabilidad de fraude y el de mayor precio final, y muestra los resultados, indicando además qué cliente debería revisarse manualmente por presentar el mayor riesgo de fraude.


```python
def analisi(clients):

    frau = major_frau(clients)

    preu = major_preu(clients)
    print("\nANÀLISI")
    print("Client amb més probabilitat de frau:", frau["nom"])
    print("Client amb el preu final més alt:", preu["nom"])
    print("Client prioritari per revisar:", frau["nom"])
```

```python
clients = [
    {"nom": "Ana", "edat": 42, "codi": 0},
    {"nom": "Carlos", "edat": 30, "codi": 2},
    {"nom": "Isabel", "edat": 55, "codi": 3},
    {"nom": "Jorge", "edat": 40, "codi": 1},
    {"nom": "Marta", "edat": 28, "codi": 0}
]

nivells_seguro = {
    0: {"categoria": "Baix risc", "preu_base": 120, "fraude": 0.01},
    1: {"categoria": "Risc mitjà", "preu_base": 200, "fraude": 0.05},
    2: {"categoria": "Risc alt", "preu_base": 350, "fraude": 0.15},
    3: {"categoria": "Risc crític", "preu_base": 500, "fraude": 0.30}
}

def classificar_client(client, nivells_seguro):

    dades = nivells_seguro[client["codi"]]

    client["categoria"] = dades["categoria"]
    client["preu_base"] = dades["preu_base"]
    client["probabilitat_frau"] = dades["fraude"]
    return client

def calcular_preu(client):

    preu = client["preu_base"] * (1 + client["edat"] / 100)

    client["preu_final"] = round(preu,2)

    return client

def processar_clients(clients, nivells_seguro):

    resultat = []

    for client in clients:
        client = classificar_client(client, nivells_seguro)
        client = calcular_preu(client)

        resultat.append(client)

    return resultat

def mostrar_informe(clients):

    print("\nINFORME")

    for client in clients:

        print("-----------------------")
        print("Nom:", client["nom"])
        print("Edat:", client["edat"])
        print("Categoria:", client["categoria"])
        print("Preu base:", client["preu_base"])
        print("Preu final:", client["preu_final"])
        print("Probabilitat frau:", client["probabilitat_frau"])


def major_frau(clients):
    major = clients[0]

    for client in clients:
        if client["probabilitat_frau"] > major["probabilitat_frau"]:
            major = client

    return major

def major_preu(clients):

    major = clients[0]

    for client in clients:
        if client["preu_final"] > major["preu_final"]:
            major = client

    return major


def ordenar_frau(clients):

    for i in range(len(clients)-1):
        for j in range(i+1, len(clients)):
            if clients[i]["probabilitat_frau"] < clients[j]["probabilitat_frau"]:
                auxiliar = clients[i]
                clients[i] = clients[j]
                clients[j] = auxiliar

    return clients


def ordenar_preu(clients):

    for i in range(len(clients)-1):
        for j in range(i+1, len(clients)):
            if clients[i]["preu_final"] < clients[j]["preu_final"]:

                auxiliar = clients[i]
                clients[i] = clients[j]
                clients[j] = auxiliar

    return clients

def analisi(clients):

    frau = major_frau(clients)

    preu = major_preu(clients)
    print("\nANÀLISI")
    print("Client amb més probabilitat de frau:", frau["nom"])
    print("Client amb el preu final més alt:", preu["nom"])
    print("Client prioritari per revisar:", frau["nom"])

def main():

    clients_final = processar_clients(clients, nivells_seguro)

    mostrar_informe(clients_final)

    print("\nOrdenats per frau")
    clients_frau = ordenar_frau(clients_final)
    mostrar_informe(clients_frau)

    print("\nOrdenats per preu")
    clients_preu = ordenar_preu(clients_final)
    mostrar_informe(clients_preu)

    analisi(clients_final)

main()


```

###**1-FUNCION obtener caracteres**
La función obtenir_caracters() crea una cadena con todos los caracteres disponibles según las opciones seleccionadas (mayúsculas, minúsculas, números y signos especiales) y la devuelve para generar la contraseña

```python
def obtenir_caracters(majuscules, minuscules, numeros, signes):

    caracters = ""

    if majuscules:
        caracters += MAJUSCULES

    if minuscules:
        caracters += MINUSCULES

    if numeros:
        caracters += NUMEROS

    if signes:
        caracters += SIGNES

    return caracters
```

###**2-FUNCION añadir caracteres obligatorios**
 La función afegir_obligatoris obliga a incluir al menos un carácter de cada tipo que el usuario haya seleccionado.

```python
def afegir_obligatoris(majuscules, minuscules, numeros, signes):

    contrasenya = []

    if majuscules:
        contrasenya.append(random.choice(MAJUSCULES))  #Si mayusculas es True, escoje una letra al azar y añadela a "contrasenya"

    if minuscules:
        contrasenya.append(random.choice(MINUSCULES))

    if numeros:
        contrasenya.append(random.choice(NUMEROS))

    if signes:
        contrasenya.append(random.choice(SIGNES))

    return contrasenya
```

###**3-FUNCION completar contraseña**
 La funcion completar_contrasenya completa la contraseña añadiendo caracteres aleatorios hasta alcanzar la longitud indicada.

```python
def completar_contrasenya(contrasenya, caracters, longitud):
    while len(contrasenya) < longitud:
      contrasenya.append(random.choice(caracters))

    return contrasenya
```

###**4-FUNCION mezclar contraseña**
 Mezcla aleatoriamente el orden de los caracteres.

```python
def barrejar_contrasenya(contrasenya):
    random.shuffle(contrasenya)    #mezclar el orden de los caracteres de la contraseña.

    return contrasenya
```

###**5-FUNCION  crear contraseña**
Es la función principal del generador.

Coordina todas las funciones anteriores.

```python
def crear_contrasenya(longitud,
                      majuscules=True,
                      minuscules=True,
                      numeros=True,
                      signes=False):

    caracters = obtenir_caracters(majuscules, minuscules, numeros,signes) #llama a la funcion para crear una cadena con todos los caracteres disponibles

    contrasenya = afegir_obligatoris(majuscules,minuscules,numeros,signes) #llama a la funcion que añade un caracter de cada tipo selecionado

    contrasenya = completar_contrasenya(contrasenya,caracters,longitud)

    contrasenya = barrejar_contrasenya(contrasenya)

    return "".join(contrasenya)
```

```python
import random

MAJUSCULES = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
MINUSCULES = "abcdefghijklmnopqrstuvwxyz"
NUMEROS = "0123456789"
SIGNES = "!@#$%&*?"

def obtenir_caracters(majuscules, minuscules, numeros, signes):

    caracters = ""

    if majuscules:
        caracters += MAJUSCULES

    if minuscules:
        caracters += MINUSCULES

    if numeros:
        caracters += NUMEROS

    if signes:
        caracters += SIGNES

    return caracters


def afegir_obligatoris(majuscules, minuscules, numeros, signes):

    contrasenya = []

    if majuscules:
        contrasenya.append(random.choice(MAJUSCULES))

    if minuscules:
        contrasenya.append(random.choice(MINUSCULES))

    if numeros:
        contrasenya.append(random.choice(NUMEROS))

    if signes:
        contrasenya.append(random.choice(SIGNES))

    return contrasenya


def completar_contrasenya(contrasenya, caracters, longitud):
    while len(contrasenya) < longitud:
      contrasenya.append(random.choice(caracters))

    return contrasenya


def barrejar_contrasenya(contrasenya):
    random.shuffle(contrasenya)

    return contrasenya


def crear_contrasenya(longitud,
                      majuscules=True,
                      minuscules=True,
                      numeros=True,
                      signes=False):

    caracters = obtenir_caracters(majuscules, minuscules, numeros,signes)

    contrasenya = afegir_obligatoris(majuscules,minuscules,numeros,signes)

    contrasenya = completar_contrasenya(contrasenya,caracters,longitud)

    contrasenya = barrejar_contrasenya(contrasenya)

    return "".join(contrasenya)


def main():

    longitud = int(input("Longitud de la contrasenya: "))

    contrasenya = crear_contrasenya(longitud)

    print(contrasenya)

main()
```

###**1-FUNCION Leer el archivo**

```python
def llegir_partits(nombre_archivo):

    with open(nombre_archivo, "r", encoding="utf-8") as fichero:
        lineas = fichero.readlines()

    return lineas

```

###**2-FUNCION. procesar todos los partidos**
 La función procesar_partits() recorre todas las líneas del archivo, extrae la información de cada partido, calcula los goles a favor, los goles en contra y los puntos de cada equipo, y devuelve un diccionario con las estadísticas finales.

```python
def procesar_partits(lineas):

    estadisticas = {}

    for linea in lineas:

        linea = linea.strip()

        datos = linea.split("\t")

        local = datos[0]
        resultado = datos[1]
        visitante = datos[2]

        goles = resultado.split("-")

        goles_local = int(goles[0])
        goles_visitante = int(goles[1])

        if local not in estadisticas:

            estadisticas[local] = {
                "goles_favor": 0,
                "goles_contra": 0,
                "puntos": 0
            }

        if visitante not in estadisticas:

            estadisticas[visitante] = {
                "goles_favor": 0,
                "goles_contra": 0,
                "puntos": 0
            }

        estadisticas[local]["goles_favor"] += goles_local
        estadisticas[local]["goles_contra"] += goles_visitante

        estadisticas[visitante]["goles_favor"] += goles_visitante
        estadisticas[visitante]["goles_contra"] += goles_local

        if goles_local > goles_visitante:

            estadisticas[local]["puntos"] += 3

        elif goles_visitante > goles_local:

            estadisticas[visitante]["puntos"] += 3

        else:

            estadisticas[local]["puntos"] += 1
            estadisticas[visitante]["puntos"] += 1

    return estadisticas
```

###**3-FUNCION equipo más goleador**
La función equip_mes_golejador() devuelve el nombre del equipo que ha marcado más goles, buscando el valor máximo de goles_favor dentro del diccionario de estadísticas.

```python
def equip_mes_golejador(estadisticas):

    return max(
        estadisticas,
        key=lambda equipo: estadisticas[equipo]["goles_favor"]
    )
```

###**4-FUNCION equipo más goleado**
La función equip mes golejat() devuelve el nombre del equipo que ha encajado más goles buscando el valor máximo de goles_contra dentro del diccionario de estadisticas.

```python
def equip_mes_golejat(estadisticas):

    return max(
        estadisticas,
        key=lambda equipo: estadisticas[equipo]["goles_contra"]
    )
```

###**5-FUNCION. Clasificación final**
La función ordena los equipos por los puntos conseguidos, de mayor a menor, utilizando sorted(), y devuelve la clasificación final para mostrar la posición de cada equipo

```python
def clasificacio_final(estadisticas):

    clasificacio = sorted(estadisticas.items(),key=lambda equipo: equipo[1]["puntos"], reverse=True)

    return clasificacio
```
