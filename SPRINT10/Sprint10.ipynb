## 3.Fes una petició GET senzilla:


>>* URL:

```python
url = 'https://api.open-meteo.com/v1/forecast'
```

>>* Parámetros para la ciudad de Barcelona:

```python
parametro_bcn = {
    'latitude': 41.3887,
    'longitude': 2.1590,
    'hourly': ['temperature_2m', 'relative_humidity_2m'],
    'timezone': 'Europe/Madrid'}
```

>>* Petición:

```python
response = requests.get(url, params = parametro_bcn)
```

### Mostra el codi d'estat de la resposta.

```python
print(f'Codigo de estado HTTP: {response.status_code}')
```

### Imprimeix de forma clara alguns dels camps de la resposta JSON.

```python
data = response.json()
```

```python
print(f'Latitud: {data['latitude']}')
print(f'Longitud: {data['longitude']}')
print(f'Elevación: {data['elevation']} metres')
print(f'Zona horaria: {data['timezone']}')

print('\nMustra de las 3 primeras horas de registro:')
for i in range(3):
    hora = data['hourly']['time'][i]
    temp = data['hourly']['temperature_2m'][i]
    humitat = data['hourly']['relative_humidity_2m'][i]
    print(f"- Fecha/Hora: {hora} | Temperatura: {temp}°C | Humedad: {humitat}%")
```

## 4.Converteix les dades a un DataFrame de pandas:



>>* Extraemos el diccionario de datos horarios.

```python
hourly_data = data['hourly']
```

>>* Convertimos el diccionario a un DataFrame.

```python
df_meteo = pd.DataFrame(hourly_data)
```

>>* Convertimos la columna d etiempo a formato datetime de Pandas.

```python
df_meteo['time'] = pd.to_datetime(df_meteo['time'])
```

# Nivell 3
API d'Open Data Barcelona

## 1.Utilitza l’API d’Open Data BCN per:


```python
BASE_URL_BCN = 'https://opendata-ajuntament.barcelona.cat/data/api/action'
```

### Cercar un dataset del teu interès mitjançant package_search i package_show. .

Utilizamos package_search para buscar datasets de 'accidentes'

```python
search_url = f'{BASE_URL_BCN}/package_search'
params_search = {'q': 'accidents'}
```

```python
res_search = requests.get(search_url, params=params_search)
print(f'Codigo de estado (package_search): {res_search.status_code}')
```

```python
data_search = res_search.json()
datasets_trobats = data_search['result']['results']
```

## 2.Dels resultats obtinguts, selecciona’n un que tingui recursos disponibles en CSV o JSON.

Utilizamos *package_show* para ver los recursos disponibles del dataset.

```python
dataset_id = datasets_trobats[0]['id']
print(f'Dataset seleccionat: {datasets_trobats[0]['title']} (ID: {dataset_id})')
```

```python
show_url = f'{BASE_URL_BCN}/package_show'
res_show = requests.get(show_url, params={'id': dataset_id})
print(f"Codigo de estado (package_show): {res_show.status_code}")

data_show = res_show.json()
resources = data_show['result']['resources']
```

# Buscamos un recurso que este en formato CSV o JSON y que tenga el dataset activo.

```python
selected_resource = None
for res in resources:
    if res.get('format', '').upper() in ['CSV', 'JSON'] and res.get('datastore_active'):
        selected_resource = res
        break
```

Si no en trobem cap amb datastore_active=True, seleccionem el primer que sigui CSV/JSON

```python
if not selected_resource:
    for res in resources:
        if res.get('format', '').upper() in ['CSV', 'JSON']:
            selected_resource = res
            break

resource_id = selected_resource['id']
print(f'Nombre del recurso seleccionado: {selected_resource['name']}')
print(f'Formato: {selected_resource['format']}')
print(f'Resource ID: {resource_id}')
```

## 3.Amb el resource_id del recurs seleccionat, realitza una consulta mitjançant datastore_search per:


###Recuperar almenys 100 registres.

```python
datastore_url = f'{BASE_URL_BCN}/datastore_search'
params_datastore = {
    'resource_id': resource_id,
    'limit': 100}
```

```python
res_datastore = requests.get(datastore_url, params=params_datastore)
print(f'Codigo de estado (datastore_search): {res_datastore.status_code}')
```

```python
data_datastore = res_datastore.json()
```

Extraemos los registros

```python
records = data_datastore['result']['records']
print(f'Número de registros recuperados: {len(records)}')
```

## 4.Converteix els resultats a un DataFrame.


```python
df_bcn = pd.DataFrame(records)
```

```python
df_bcn.head()
```

## 5.Desa el DataFrame en un fitxer .csv.

```python
nom_fitxer = 'opendata_bcn_resultats.csv'
df_bcn.to_csv(nom_fitxer, index=False, encoding='utf-8')
```
